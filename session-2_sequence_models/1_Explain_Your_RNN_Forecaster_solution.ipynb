## 0. Setup

```python
import time
import numpy as np, pandas as pd, matplotlib.pyplot as plt
import torch, torch.nn as nn
from scipy.stats import spearmanr

SEED = 42
np.random.seed(SEED); torch.manual_seed(SEED)
plt.rcParams.update({"axes.spines.top": False, "axes.spines.right": False, "figure.dpi": 110})
BLUE, RED, GREY = "#2563EB", "#DC2626", "#9CA3AF"
print("torch", torch.__version__, "| the models are small, so this lab runs on the CPU")
```

## 1. The same data and the same models as the earlier lab · 🟢 core

`generate_time_series` is copied from the earlier notebook: every series is **two sine waves plus a little noise**. The model sees steps 0–49 and predicts step 50.

The two models are the earlier lab's *linear baseline* (`Flatten` + `Dense(1)`) and its *deep RNN with LSTM cells* (two layers of 20 units), written in PyTorch.

```python
def generate_time_series(batch_size, n_steps):
    freq1, freq2, offsets1, offsets2 = np.random.rand(4, batch_size, 1)
    time_ = np.linspace(0, 1, n_steps)
    series = 0.5 * np.sin((time_ - offsets1) * (freq1 * 10 + 10))   #   wave 1
    series += 0.2 * np.sin((time_ - offsets2) * (freq2 * 20 + 20))  # + wave 2
    series += 0.1 * (np.random.rand(batch_size, n_steps) - 0.5)     # + noise
    return series[..., np.newaxis].astype(np.float32)

n_steps = 50
series = torch.tensor(generate_time_series(10000, n_steps + 1))
X_train, y_train = series[:7000, :n_steps], series[:7000, -1, 0]
X_valid, y_valid = series[7000:9000, :n_steps], series[7000:9000, -1, 0]
print("X_train", tuple(X_train.shape), "(series, time steps, 1 feature)   y_train", tuple(y_train.shape))

class LinearForecaster(nn.Module):                 # Flatten + Dense(1)
    def __init__(self):
        super().__init__(); self.out = nn.Linear(n_steps, 1)
    def forward(self, x): return self.out(x.flatten(1)).squeeze(-1)

class LSTMForecaster(nn.Module):                   # LSTM(20) -> LSTM(20) -> Dense(1) on the last state
    def __init__(self, units=20):
        super().__init__(); self.rnn = nn.LSTM(1, units, num_layers=2, batch_first=True); self.out = nn.Linear(units, 1)
    def forward(self, x): return self.out(self.rnn(x)[0][:, -1]).squeeze(-1)

def train(model, epochs=20, lr=3e-3, batch=64):
    opt = torch.optim.Adam(model.parameters(), lr); t = time.time()
    for _ in range(epochs):
        for b in torch.randperm(len(X_train)).split(batch):
            loss = nn.functional.mse_loss(model(X_train[b]), y_train[b]); opt.zero_grad(); loss.backward(); opt.step()
    model.eval()
    for p in model.parameters(): p.requires_grad_(False)       # from now on we only need gradients with respect to the input
    print(f"{type(model).__name__}: trained in {time.time() - t:.0f} s")
    return model

linear, lstm = train(LinearForecaster()), train(LSTMForecaster())
MODELS = {"Linear": linear, "LSTM": lstm}

@torch.no_grad()
def mse(model, X=X_valid, y=y_valid): return float(((model(X) - y) ** 2).mean())
display(pd.DataFrame({"Validation MSE": {"Naive (repeat the last value)": float(((X_valid[:, -1, 0] - y_valid) ** 2).mean()),
                                         "Linear": mse(linear), "LSTM": mse(lstm)}}).round(4))
```

## 2. Integrated Gradients by hand · 🟢 core

IG explains the **difference** between the forecast for the series $x$ and the forecast for a **baseline** series $x'$ (here: a flat line at zero). Each time step $i$ gets

$$\mathrm{IG}_i = (x_i - x'_i) \times \frac{1}{m} \sum_{k=1}^{m} \frac{\partial f}{\partial x_i}\Big(x' + \alpha_k (x - x')\Big), \qquad \alpha_k = \frac{k - 0.5}{m}$$

In words: **walk** from the baseline to the series in $m$ small steps, **average the gradient** along the way, and **multiply** by how far each input moved.

Two facts let us check the code:

- **Linear model.** The gradient is the weight $w_i$ everywhere, so $\mathrm{IG}_i = w_i (x_i - x'_i)$ exactly.
- **Completeness.** For any model the attributions add up to $f(x) - f(x')$.

```python
def integrated_gradients(f, x, baseline, steps=64):
    # f: maps a batch (B, T, C) to B numbers. x, baseline: one series each, shape (T, C). Returns attributions, shape (T, C).
    alphas = (torch.arange(steps, device=x.device) + 0.5) / steps                 # the m points on the path
    path = baseline + alphas.view(-1, 1, 1) * (x - baseline)                      # (steps, T, C): all points on the path, as one batch
    path.requires_grad_(True)
    grads = torch.autograd.grad(f(path).sum(), path)[0]                           # gradient of the output at every point on the path
    return ((x - baseline) * grads.mean(0)).detach()                              # average gradient x distance moved
```

### ✅ Check: `integrated_gradients`

```python
x, zero = X_valid[0], torch.zeros_like(X_valid[0])
a = integrated_gradients(linear, x, zero)
assert a.shape == x.shape, f"attributions must have the shape of the input {tuple(x.shape)}, got {tuple(a.shape)}"
exact = linear.out.weight[0] * x[:, 0]
assert torch.allclose(a[:, 0], exact, atol=1e-5), "on the linear model IG must equal weight x input"
a = integrated_gradients(lstm, x, zero, steps=256)
gap = float(lstm(x[None]) - lstm(zero[None]))
assert abs(float(a.sum()) - gap) < 0.01, f"completeness fails on the LSTM: sum {float(a.sum()):.4f} vs f(x) - f(x') {gap:.4f}"
print("✅ integrated_gradients is right: exact on the linear model, complete on the LSTM")
```

## 3. Completeness and the number of steps · 🟢 core

The sum over the path only approximates an integral. How many steps does this small LSTM need before the attributions add up?

```python
rows = []
for m in [1, 2, 4, 8, 16, 32, 64, 256]:
    a = integrated_gradients(lstm, x, zero, steps=m)
    rows.append({"steps": m, "sum of attributions": float(a.sum()), "f(x) - f(x')": gap, "error": abs(float(a.sum()) - gap)})
display(pd.DataFrame(rows).round(4))
```

## 4. Which time steps does the forecast use? · 🟢 core

**One forecast first.** The top row shows three validation series, the true next value (cross) and the LSTM forecast (dot). The bottom row shows the IG attribution of every time step: red pushes the forecast up, blue pushes it down.

```python
fig, axes = plt.subplots(2, 3, figsize=(13, 5.2), sharex=True)
for j in range(3):
    x = X_valid[j]; a = integrated_gradients(lstm, x, torch.zeros_like(x))[:, 0].numpy()
    axes[0, j].plot(x[:, 0], ".-", color=GREY); axes[0, j].plot(n_steps, y_valid[j], "kx", ms=9, label="true next value")
    axes[0, j].plot(n_steps, float(lstm(x[None])), "o", color=RED, label="LSTM forecast"); axes[0, j].set_title(f"validation series {j}", fontsize=10)
    axes[1, j].bar(range(n_steps), a, color=np.where(a > 0, RED, BLUE)); axes[1, j].axhline(0, color="k", lw=0.5); axes[1, j].set_xlabel("time step")
axes[0, 0].legend(frameon=False, fontsize=8); axes[0, 0].set_ylabel("series"); axes[1, 0].set_ylabel("IG attribution")
plt.tight_layout(); plt.show()
```

**Now many forecasts.** One series proves little. Average the size of the attribution, $|\mathrm{IG}_i|$, over 300 validation series, for both models. Each curve is scaled to add up to 1, so it reads as *the share of the explanation that falls on each time step*.

```python
N_EXPLAIN = 300
def mean_abs_ig(model, baseline_fn=torch.zeros_like):
    A = torch.stack([integrated_gradients(model, x, baseline_fn(x), steps=32)[:, 0].abs() for x in X_valid[:N_EXPLAIN]])
    return (A.mean(0) / A.mean(0).sum()).numpy()
profile = {name: mean_abs_ig(m) for name, m in MODELS.items()}

fig, ax = plt.subplots(figsize=(9, 3.6))
ax.plot(profile["Linear"], "o-", ms=3, color=GREY, label="Linear"); ax.plot(profile["LSTM"], "o-", ms=3, color=BLUE, label="LSTM")
ax.set_xlabel("time step (49 = the most recent value)"); ax.set_ylabel("share of |IG|"); ax.legend(frameon=False); plt.tight_layout(); plt.show()
display(pd.DataFrame({name: {f"share of |IG| on the last {k} steps": p[-k:].sum() for k in [1, 3, 5, 10, 25]} for name, p in profile.items()}).round(2))
```

## 5. The baseline matters · ⚪ if time

IG explains the forecast **relative to a baseline**. A flat line at zero is one choice of "no information". Two others are just as defensible:

| Baseline | What it means |
|---|---|
| `zero` | a flat line at 0, the long-run average of every series |
| `mean` | a flat line at this series' own average |
| `last` | a flat line at this series' last value: "nothing changed" |

Completeness holds for every baseline, but the attributions are different numbers, because the question is different.

```python
BASELINES = {"zero": torch.zeros_like, "mean": lambda x: torch.full_like(x, float(x.mean())), "last": lambda x: torch.full_like(x, float(x[-1, 0]))}
x = X_valid[0]; attr = {}
fig, axes = plt.subplots(1, 3, figsize=(13, 2.8), sharey=True)
for ax, (name, fn) in zip(axes, BASELINES.items()):
    b = fn(x); attr[name] = integrated_gradients(lstm, x, b)[:, 0].numpy()
    ax.bar(range(n_steps), attr[name], color=np.where(attr[name] > 0, RED, BLUE)); ax.axhline(0, color="k", lw=0.5)
    ax.set_title(f"baseline: {name}    f(x) - f(x') = {float(lstm(x[None]) - lstm(b[None])):+.2f}", fontsize=10); ax.set_xlabel("time step")
axes[0].set_ylabel("IG attribution"); plt.tight_layout(); plt.show()
print("Rank correlation between the three explanations of the same forecast:")
display(pd.DataFrame([[spearmanr(attr[a], attr[b])[0] for b in BASELINES] for a in BASELINES], index=list(BASELINES), columns=list(BASELINES)).round(2))
```

## 6. Occlusion over time windows: the model-agnostic view · ⚪ if time

IG needs gradients, so it only works when we can open the model. **Occlusion** needs nothing but forecasts: hide a **window** of time steps (set it to the fill value), run the model again, and record how far the forecast moves.

$$\text{drop}_j = f(x) - f(x \text{ with window } j \text{ hidden})$$

This is the occlusion of the EuroSAT lab, with a window in time in place of a patch in the image.

To compare with IG, add up the IG attributions inside each window. On the **linear** model the two must agree exactly: hiding a window removes $\sum w_i x_i$ over that window, which is the IG sum.

```python
@torch.no_grad()
def occlusion_time(f, x, window=5, fill=0.0):
    # Hide one window of time steps at a time. x: (T, C). Returns the drop for each of the T // window windows.
    n = x.shape[0] // window
    batch = x.repeat(n, 1, 1)                                   # n copies of the series
    for j in range(n):
        batch[j, j * window:(j + 1) * window] = fill            # copy j has window j hidden
    return (f(x[None]) - f(batch)).numpy()
```

### ✅ Check: `occlusion_time`

```python
x, zero = X_valid[0], torch.zeros_like(X_valid[0])
d = occlusion_time(linear, x, window=5)
assert d.shape == (10,), f"50 steps and a window of 5 give 10 drops, got shape {d.shape}"
ig_windows = integrated_gradients(linear, x, zero)[:, 0].view(10, 5).sum(1).numpy()
assert np.allclose(d, ig_windows, atol=1e-5), "on the linear model the occlusion drop of a window must equal the IG sum over that window"
whole = occlusion_time(lstm, x, window=50)
assert np.allclose(whole, float(lstm(x[None]) - lstm(zero[None])), atol=1e-5), "hiding everything must give f(x) - f(flat line at the fill value)"
print("✅ occlusion_time is right")
```

```python
W = 5; N_COMPARE = 200
occ = np.stack([occlusion_time(lstm, x, W) for x in X_valid[:N_COMPARE]])
igw = np.stack([integrated_gradients(lstm, x, torch.zeros_like(x), steps=32)[:, 0].view(-1, W).sum(1).numpy() for x in X_valid[:N_COMPARE]])
rho = np.mean([spearmanr(o, g)[0] for o, g in zip(occ, igw)])

fig, axes = plt.subplots(1, 2, figsize=(12, 3.6))
x_pos = np.arange(n_steps // W)
axes[0].bar(x_pos - 0.2, np.abs(occ).mean(0), 0.4, color=GREY, label="occlusion: mean |drop|")
axes[0].bar(x_pos + 0.2, np.abs(igw).mean(0), 0.4, color=BLUE, label="IG: mean |sum over the window|")
axes[0].set_xticks(x_pos); axes[0].set_xticklabels([f"{j * W}-{(j + 1) * W - 1}" for j in x_pos], fontsize=8); axes[0].set_xlabel("window of time steps")
axes[0].legend(frameon=False, fontsize=9)
axes[1].scatter(igw.ravel(), occ.ravel(), s=4, alpha=0.4, color=BLUE); lim = np.abs(occ).max() * 1.05
axes[1].plot([-lim, lim], [-lim, lim], color=GREY, lw=1); axes[1].set_xlabel("IG, summed over the window"); axes[1].set_ylabel("occlusion drop")
axes[1].set_title(f"LSTM, {N_COMPARE} series x {n_steps // W} windows", fontsize=10)
plt.tight_layout(); plt.show()
print(f"Mean rank correlation between occlusion and IG, per series: {rho:.2f}")
```

## 7. Test the explanation: keep only the last *k* steps · 🟢 core

IG put most of the attribution on the last few steps and very little on the first half of the series (the table in section 4). It is tempting to conclude that the model **does not need** the early steps. A claim about the model can be tested **on the model**:

- **Keep the last *k* steps** and flatten everything before them to zero. If the early steps are not needed, the error stays low for a small *k*.
- **Hide the last *k* steps** and keep everything before them. If the last steps carry the forecast, the error jumps at once.

This is the deletion test of the method-comparison lab, along the time axis. Before you run it, guess: *with only the last 10 steps, how much worse is the LSTM?*

```python
def keep_last(X, k, fill=0.0):
    # A copy of the batch X (N, T, C) in which only the last k time steps survive.
    Xk = X.clone(); Xk[:, :X.shape[1] - k] = fill
    return Xk

def hide_last(X, k, fill=0.0):
    # A copy of the batch X (N, T, C) in which the last k time steps are hidden.
    Xk = X.clone(); Xk[:, X.shape[1] - k:] = fill
    return Xk
```

### ✅ Check: `keep_last` and `hide_last`

```python
T_ = torch.ones(2, 50, 1)
assert keep_last(T_, 3).sum() == 6 and keep_last(T_, 3)[0, -3:].sum() == 3, "keep_last(X, 3) must keep exactly the last 3 steps"
assert hide_last(T_, 3).sum() == 94 and hide_last(T_, 3)[0, -3:].sum() == 0, "hide_last(X, 3) must hide exactly the last 3 steps"
assert torch.equal(keep_last(T_, 50), T_) and torch.equal(hide_last(T_, 0), T_) and T_.sum() == 100, "k = 50 keeps everything, k = 0 hides nothing, and X itself must not change"
print("✅ keep_last and hide_last are right")
```

```python
ks = [0, 1, 2, 3, 5, 10, 20, 30, 50]
keep = pd.DataFrame({name: [mse(m, keep_last(X_valid, k)) for k in ks] for name, m in MODELS.items()}, index=ks)
hide = pd.DataFrame({name: [mse(m, hide_last(X_valid, k)) for k in ks] for name, m in MODELS.items()}, index=ks)

fig, axes = plt.subplots(1, 2, figsize=(12, 3.6), sharey=True)
for ax, df, title in zip(axes, [keep, hide], ["Keep only the last k steps", "Hide the last k steps"]):
    ax.plot(df.index, df["Linear"], "o-", color=GREY, label="Linear"); ax.plot(df.index, df["LSTM"], "o-", color=BLUE, label="LSTM")
    ax.set_yscale("log"); ax.set_xlabel("k"); ax.set_title(title, fontsize=10)
axes[0].set_ylabel("validation MSE (log scale)"); axes[0].legend(frameon=False); plt.tight_layout(); plt.show()
display(pd.concat({"keep the last k": keep, "hide the last k": hide}, axis=1).round(4))
```
