PCI gives one number per band for the whole test set. LIME (lab 2) gives one band importance per image, but it is a regression fit with no
guarantees. **Shapley values** split one prediction fairly among the players, with properties LIME lacks:

$$\phi_j(x) = \sum_{S \subseteq N \setminus \{j\}} \frac{|S|!\,(n - |S| - 1)!}{n!}\,\big[v(S \cup \{j\}) - v(S)\big]$$

* **Players:** the $n = 14$ input channels (13 bands + the NOISE control).
* **Value function $v(S)$:** $p(\text{class})$ when only the channels in $S$ are kept and the others are set to their training mean.
* **Efficiency:** $\sum_j \phi_j = v(N) - v(\varnothing) = p(\text{class} \mid x) - p(\text{class} \mid \text{all bands at the mean})$.

With 14 players there are only $2^{14} = 16{,}384$ coalitions, so we can compute **exact** Shapley values for one image and check KernelSHAP
against them. Then we explain hundreds of images with KernelSHAP and aggregate: **beeswarm** (all local explanations at once), **mean |SHAP|**
(a global ranking to compare with PCI), per-class summaries, and grouped players for correlated bands.

**What you should see.** KernelSHAP matches the exact values closely with far fewer model calls. The global SHAP ranking broadly agrees with PCI
but not perfectly, because the two methods remove bands differently and measure different things. NOISE gets ≈ 0.

## 0. The model and the images: load the lab pack

**This is setup. Run the next cell and move on; you do not need to read it to follow the lab.**

It downloads a small *lab pack* (about 40 MB, a few seconds) that holds a CNN already trained on the EuroSAT satellite images and 500 of
its test images (50 per class). Nothing is trained in class. Every image has 14 channels: the 13 Sentinel-2 bands plus one channel of pure
noise that the model should ignore.

```python
# ---- the model: a small CNN ----
class SmallCNN(nn.Module):
    def __init__(self, in_ch, n_classes=10, w=48):
        super().__init__()

        # One block = two times (3 x 3 convolution, batch norm, ReLU).
        def block(i, o):
            return nn.Sequential(
                nn.Conv2d(i, o, 3, padding=1, bias=False),
                nn.BatchNorm2d(o),
                nn.ReLU(inplace=True),
                nn.Conv2d(o, o, 3, padding=1, bias=False),
                nn.BatchNorm2d(o),
                nn.ReLU(inplace=True),
            )

        # Four blocks; after each of the first three the image is halved (64 ->
        # 32 -> 16 -> 8 pixels).
        self.features = nn.Sequential(
            block(in_ch, w),
            nn.MaxPool2d(2),
            block(w, 2 * w),
            nn.MaxPool2d(2),
            block(2 * w, 4 * w),
            nn.MaxPool2d(2),
            block(4 * w, 4 * w),
        )
        # Average every feature map to one number, then one linear layer gives
        # the 10 class scores.
        self.head = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Dropout(0.3),
            nn.Linear(4 * w, n_classes),
        )

    def forward(self, x):
        return self.head(self.features(x))


# ---- download the lab pack (skipped if the file is already here, e.g. uploaded
# by hand)
PACK_FILE = Path("eurosat_lab_pack_v1.npz")
for url in LAB_PACK_URLS:
    if PACK_FILE.exists():
        break
    try:
        print(f"downloading the lab pack from {url} ...")
        urllib.request.urlretrieve(url, PACK_FILE)
    except Exception as e:
        print(f"  failed: {type(e).__name__}: {e}")
        PACK_FILE.unlink(missing_ok=True)
if not PACK_FILE.exists():
    raise RuntimeError(
        "Could not download the lab pack. Drag eurosat_lab_pack_v1.npz into the Colab Files panel and run this cell again."
    )
pack = np.load(PACK_FILE)

# ---- the trained model: put the saved weights into the network ----
model = SmallCNN(N_CH).to(DEVICE)
model.load_state_dict(
    {
        k[3:]: torch.from_numpy(pack[k])
        for k in pack.files
        if k.startswith("w::")
    }
)
# evaluation mode: no dropout, fixed batch-norm statistics
model.eval()

# ---- the test images ----
# Standardise every band with the mean and standard deviation of the training
# data.
# After this step the value 0 means "the average value of this band", which is
# how we will "remove" things later.
band_mean, band_std = pack["band_mean"], pack["band_std"]
x = (
    pack["X_test_raw"].astype(np.float32)
    - band_mean[None, :, None, None].astype(np.float32)
) / band_std[None, :, None, None].astype(np.float32)
# the true class of every test image
y_test = pack["y_test"].astype(np.int64)

# Add the noise channel: one fixed noise image per test image, so the model
# always gives the same answer for the same image.
NOISE = torch.randn(
    len(x), 1, 64, 64, generator=torch.Generator().manual_seed(SEED)
).half()
# X has shape (500, 14, 64, 64)
X = torch.cat([torch.from_numpy(x).half(), NOISE], 1)
del x, pack
print("test images:", tuple(X.shape))
```

### Prediction helpers and example images
`coalition_proba` is the one perturbation used by every method in these labs: *remove* a set of pixels or channels by setting them to the training mean, and read the probability of the class we explain.

```python
@torch.no_grad()
def predict_proba(x, batch=EVAL_BATCH):
    # Class probabilities for a (N, 14, H, W) or (14, H, W) tensor, as a numpy
    # array (N, 10).
    if x.ndim == 3:
        x = x[None]
    return torch.cat(
        [
            F.softmax(model(x[s : s + batch].to(DEVICE).float()), 1).cpu()
            for s in range(0, len(x), batch)
        ]
    ).numpy()


P = predict_proba(X)
pred = P.argmax(1)
print(
    f"Test accuracy: {(pred == y_test).mean():.4f}   (the PCI notebook reports the same model's baseline)"
)


def rgb(x):
    # True-colour composite (B04, B03, B02) of a standardised image, for display
    # only.
    x = (
        x[:13].float().numpy() * band_std[:, None, None]
        + band_mean[:, None, None]
    )
    img = np.stack([x[3], x[2], x[1]], -1)
    lo, hi = np.percentile(img, 2), np.percentile(img, 98)
    return np.clip((img - lo) / (hi - lo + 1e-6), 0, 1)


def show_map(ax, x, heat, title="", alpha=0.55, lim=None):
    # Overlay a signed attribution map on the true-colour image: red = supports
    # the class, blue = against it.
    # `lim` fixes the colour scale (to compare panels); by default each panel
    # uses its own maximum.
    ax.imshow(rgb(x))
    lim = (np.abs(heat).max() if lim is None else lim) + 1e-8
    im = ax.imshow(heat, cmap="RdBu_r", vmin=-lim, vmax=lim, alpha=alpha)
    ax.set_title(title, fontsize=8)
    ax.axis("off")
    ax.grid(False)
    return im


def coalition_proba(x, Z, target, segments=None, chunk=256):
    # p(target) when only the "players" switched on in each row of Z are kept
    # and the rest are set to the
    # training mean (0 after standardisation). Players are superpixels if
    # `segments` is given, else the 14 channels.
    x = x.float()
    out = []
    seg = (
        None
        if segments is None
        else torch.as_tensor(segments, dtype=torch.long)
    )
    for s in range(0, len(Z), chunk):
        z = torch.as_tensor(np.asarray(Z[s : s + chunk]), dtype=torch.float32)
        m = z[:, seg][:, None] if seg is not None else z[:, :, None, None]
        out.append(predict_proba(x[None] * m)[:, target])
    return np.concatenate(out)


# Example images: per class, a correctly classified test image of MEDIAN
# confidence
# (very confident images saturate near p = 1, which hides the effect of a
# perturbation).
EXAMPLES = {}
for ci, cls in enumerate(CLASSES):
    ok = np.where((y_test == ci) & (pred == ci))[0]
    if len(ok) == 0:
        ok = np.where(y_test == ci)[0]
    ok = ok[np.argsort(P[ok, ci])]
    EXAMPLES[cls] = int(ok[len(ok) // 2])
# the test image with the most probable runner-up class
AMBIGUOUS = int(np.argmax(np.sort(P, 1)[:, -2]))

fig, axes = plt.subplots(1, 11, figsize=(17, 2.2))
for ax, (cls, i) in zip(
    axes, list(EXAMPLES.items()) + [("ambiguous", AMBIGUOUS)]
):
    top = np.argsort(P[i])[::-1][:2]
    ax.imshow(rgb(X[i]))
    ax.axis("off")
    title = (
        f"ambiguous\n{CLASSES[top[0]]} {P[i, top[0]]:.2f}\n{CLASSES[top[1]]} {P[i, top[1]]:.2f}"
        if cls == "ambiguous"
        else f"{cls}\np = {P[i, CLASSES.index(cls)]:.2f}"
    )
    ax.set_title(title, fontsize=7)
plt.suptitle("Example test images used in this lab (true colour)")
plt.tight_layout()
plt.show()
```

## 1. Exact Shapley values for one image · 🟢 core

Enumerate all $2^{14}$ coalitions, evaluate $v(S)$ for each (one batched forward pass per coalition), and apply the formula above.

```python
from math import factorial
import shap


def all_coalitions(n):
    m = np.arange(2**n)
    # bitmask id, and the (2^n, n) on/off matrix
    return m, (m[:, None] >> np.arange(n)) & 1


def shapley_from_values(v, n):
    # Shapley values of n players from the values v[S] of all 2^n coalitions (S
    # = bitmask: bit j set = player j in S).
    ids = np.arange(2**n)
    size = np.array([bin(S).count("1") for S in ids])
    # weight of a coalition of size s
    wt = np.array(
        [factorial(s) * factorial(n - s - 1) / factorial(n) for s in range(n)]
    )
    phi = np.zeros(n)
    for j in range(n):
        # coalitions without player j
        S = ids[((ids >> j) & 1) == 0]
        phi[j] = (wt[size[S]] * (v[S | (1 << j)] - v[S])).sum()
    return phi


def exact_shapley(x, target, groups=None):
    # Exact Shapley values. Players = channels, or groups of channels (list of
    # channel-index lists).
    groups = [[j] for j in range(N_CH)] if groups is None else groups
    n = len(groups)
    G = np.zeros((n, N_CH), int)
    for g, chans in enumerate(groups):
        G[g, chans] = 1
    ids, Zg = all_coalitions(n)
    # value of every coalition
    v = coalition_proba(x, Zg @ G, target, chunk=EVAL_BATCH)
    # values, v(empty), v(all)
    return shapley_from_values(v, n), v[0], v[-1]


def kernel_shap(x, target, segments=None, nsamples=NSAMPLES, seed=0):
    # KernelSHAP from the shap package, with the same value function. Players =
    # channels or superpixels.
    d = N_CH if segments is None else int(segments.max()) + 1
    np.random.seed(seed)
    ex = shap.KernelExplainer(
        lambda Z: coalition_proba(x, Z, target, segments), np.zeros((1, d))
    )
    sv = ex.shap_values(
        np.ones((1, d)), nsamples=nsamples, l1_reg=False, silent=True
    )
    return np.asarray(sv, dtype=float).reshape(-1)[:d], float(
        np.asarray(ex.expected_value).reshape(-1)[0]
    )
```

### ✅ Check: `shapley_from_values` on games with known answers

```python
n = 3
ids = np.arange(2**n)
size = np.array([bin(S).count("1") for S in ids])
a = np.array([0.5, -1.0, 2.0])
additive = np.array([sum(a[j] for j in range(n) if (S >> j) & 1) for S in ids])
assert np.allclose(
    shapley_from_values(additive, n), a
), "additive game: each player's Shapley value is its own a_j"
assert np.allclose(
    shapley_from_values(size.astype(float) ** 2, n), [3, 3, 3]
), "symmetric game v = |S|^2: equal shares of v(N) = 9"
# player 0 + (player 1 or 2)
glove = np.array([1.0 if ((S & 1) and (S & 6)) else 0.0 for S in ids])
assert np.allclose(
    shapley_from_values(glove, n), [2 / 3, 1 / 6, 1 / 6]
), "glove game"
print("✅ Shapley values are right")
```

## 2. One prediction: the waterfall plot · 🟢 core

Start at $E[f] = v(\varnothing)$ (every band at its training mean), add one band's contribution at a time, end at the model's output for this
image. The grey numbers are the band's mean standardised value in this image (0 = an average pixel, positive = brighter than average).

```python
def band_features(x):
    # mean standardised value of every channel
    return x.float().mean((-2, -1)).numpy()


phi = ex_phi if RUN_EXACT else ks
shap.plots.waterfall(
    shap.Explanation(
        values=phi,
        base_values=base,
        data=band_features(X[i0]).round(2),
        feature_names=CHANNEL_NAMES,
    ),
    max_display=N_CH,
)
```

## 3. Explain many images · 🟢 core

`N_EXPLAIN` test images, the same number per class. For each we explain $p(\text{true class})$ with KernelSHAP over the 14 channels.

```python
rng = np.random.default_rng(SEED)
per_class = max(1, N_EXPLAIN // len(CLASSES))
IDX = np.concatenate(
    [
        rng.choice(
            np.where(y_test == ci)[0],
            size=min(per_class, (y_test == ci).sum()),
            replace=False,
        )
        for ci in range(len(CLASSES))
    ]
)
SV, BASE = np.zeros((len(IDX), N_CH)), np.zeros(len(IDX))
t = time.time()
for k, i in enumerate(tqdm(IDX, desc="KernelSHAP")):
    SV[k], BASE[k] = kernel_shap(X[i], int(y_test[i]))
# per-image mean value of every channel
FEAT = X[IDX].float().mean((-2, -1)).numpy()
Y = y_test[IDX]
print(f"{len(IDX)} images, {time.time() - t:.0f} s")
```

## 5. Global ranking: mean |SHAP| vs PCI · 🟢 core

Average the absolute SHAP values over all explained images and put them next to PCI on the same model. PCI is recomputed here on the full test
set with the same code as the PCI notebook (pixel shuffle, $k = 3$) and with mean fill, the removal SHAP uses.

They answer different questions: PCI = *how much accuracy is lost* if a band's spatial information is destroyed; mean |SHAP| = *how much a
band moves the probability* of the true class, on average, with credit shared fairly among bands that work together.

```python
def pci_scores(mode="shuffle", k=3):
    # PCI on the test set: accuracy drop when one channel is shuffled within
    # each image (or set to its training mean).
    base_acc = (pred == y_test).mean()
    out = {}
    for c, name in enumerate(tqdm(CHANNEL_NAMES, desc=f"PCI ({mode})")):
        accs = []
        for r in range(k if mode == "shuffle" else 1):
            g = torch.Generator().manual_seed(r)
            preds = []
            for s in range(0, len(X), EVAL_BATCH):
                xb = X[s : s + EVAL_BATCH].float().clone()
                B, _, H, W = xb.shape
                if mode == "shuffle":
                    perm = torch.argsort(torch.rand(B, H * W, generator=g), 1)
                    xb[:, c] = torch.gather(
                        xb[:, c].reshape(B, -1), 1, perm
                    ).reshape(B, H, W)
                else:
                    xb[:, c] = 0
                preds.append(predict_proba(xb).argmax(1))
            accs.append((np.concatenate(preds) == y_test).mean())
        out[name] = base_acc - np.mean(accs)
    return pd.Series(out)


ranking = pd.DataFrame({"mean |SHAP|": np.abs(SV).mean(0)}, index=CHANNEL_NAMES)
ranking["PCI (shuffle)"] = pci_scores("shuffle")
ranking["PCI (mean fill)"] = pci_scores("mean")
ranking = ranking.sort_values("mean |SHAP|", ascending=False)

fig, axes = plt.subplots(1, 3, figsize=(15, 3.8))
for ax, col in zip(axes, ranking.columns):
    ax.bar(
        ranking.index,
        ranking[col],
        color=[GROUP_COLORS[GROUP_OF[ch]] for ch in ranking.index],
    )
    ax.set_title(col)
    ax.tick_params(axis="x", rotation=90)
    ax.axhline(0, color="#52514e", lw=0.8)
axes[0].set_ylabel("Δ p(true class)")
axes[1].set_ylabel("accuracy drop")
axes[2].set_ylabel("accuracy drop")
plt.suptitle("Bands in mean |SHAP| order")
plt.tight_layout()
plt.show()
print("Spearman rank correlation between the global rankings:")
display(ranking.corr(method="spearman").round(2))
```

## 7. Dependence plot · ⚪ if time

## 8. Correlated bands: grouped players · ⚪ if time

B08 and B8A both measure near-infrared and are almost perfectly correlated. When either one is present, the model has its NIR information, so
the Shapley value **splits** the credit between them. PCI showed the same effect (E4: grouped PCI). Treating spectral groups as **one player
each** (6 players, $2^6 = 64$ coalitions: exact and cheap) gives the group's credit directly.

Compare the group Shapley value with the sum of its bands' individual Shapley values.

```python
gnames = [g for g in GROUPS if GROUPS[g]]
GSV = np.array(
    [
        exact_shapley(X[i], int(y_test[i]), groups=[GROUPS[g] for g in gnames])[
            0
        ]
        for i in tqdm(IDX, desc="group Shapley")
    ]
)
cmp = pd.DataFrame(
    {
        "group as one player": np.abs(GSV).mean(0),
        "sum of its bands (|sum| per image)": [
            np.abs(SV[:, GROUPS[g]].sum(1)).mean() for g in gnames
        ],
    },
    index=gnames,
)
ax = cmp.plot.bar(figsize=(7.5, 3.6), color=["#2a78d6", "#86b6ef"], rot=0)
ax.set_ylabel("mean |Shapley value|")
ax.set_title("Spectral groups: grouped vs individual players")
plt.tight_layout()
plt.show()
display(cmp.round(4))
```
